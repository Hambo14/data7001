# Combined Data

```python
import pandas as pd
import numpy as np

# import datasets

PROCESSED_CSV_LOC = "../processed-csvs/"

economic_df = pd.read_csv(PROCESSED_CSV_LOC + "economic_data.csv")
visiting_visas_df = pd.read_csv(PROCESSED_CSV_LOC + "visiting_visas_by_state.csv")
short_term_arrivals_df = pd.read_csv(PROCESSED_CSV_LOC + "short_term_arrivals.csv")
bitre_df = pd.read_csv("../raw-csvs/bitre/seat_utilisation.csv")
bitre_df.set_index("MONTH", inplace=True)
```

```python
# Inspect the data
economic_df.head()
```

```python
short_term_arrivals_df.head()
```

```python
# get rid of the first column in economic_df
economic_df = economic_df.iloc[:,1:]
```

## Create First Dataset on Australia Arrival Numbers

```python
# Sum visiting visas state numbers to get total numbers each month
visiting_visas_df = visiting_visas_df.groupby("year_month", as_index=False)["arrivals"].sum()

# convert year_month to date column
visiting_visas_df["Date"] = pd.to_datetime(visiting_visas_df["year_month"], format="%Y-%m")
visiting_visas_df.drop(columns='year_month', inplace=True)

# Convert short term arrivals to date time as well
short_term_arrivals_df["Date"] = pd.to_datetime(short_term_arrivals_df["Date"], format='%Y-%m-%d')
short_term_arrivals_df.head()
```

```python
# Combine with short term arrivals data

combined_df = visiting_visas_df.merge(short_term_arrivals_df, on="Date", how="left")
combined_df.head()
```

```python
bitre_df.head()
```

```python
combined_df = combined_df.merge(bitre_df, on="Date", how="left")
combined_df.head()
```

```python
# Remove state columns from dataset
combined_df.columns = [column.strip() for column in combined_df.columns]
combined_df.drop(columns=[
    'NSW','Vic','Qld','SA','WA','Tas',
    'NT','ACT','Other Territories', 'Total (State of residence/stay)'
],inplace=True)
```

```python
combined_df.columns
```

```python
# Melt to long format
combined_long_df = pd.melt(combined_df, id_vars=["Date"], value_vars=['New Zealand',
       'New Caledonia', 'Papua New Guinea', 'Vanuatu', 'Fiji',
       'Other Oceania and Antarctica', 'Total Oceania and Antarctica',
       'UK, CIs & IOM', 'Ireland', 'Austria', 'Belgium', 'France', 'Germany',
       'Netherlands', 'Switzerland', 'Denmark', 'Finland', 'Norway', 'Sweden',
       'Other North-West Europe', 'Total North-West Europe', 'Italy', 'Spain',
       'Poland', 'Russian Federation', 'Other Southern and Eastern Europe',
       'Total Southern and Eastern Europe', 'Iran', 'Israel', 'Saudi Arabia',
       'United Arab Emirates', 'Other North Africa and the Middle East',
       'Total North Africa and the Middle East', 'Cambodia', 'Thailand',
       'Vietnam', 'Indonesia', 'Malaysia', 'Philippines', 'Singapore',
       'Other South-East Asia', 'Total South-East Asia', 'China', 'Hong Kong',
       'Taiwan', 'Japan', 'Korea, South', 'Other North-East Asia',
       'Total North-East Asia', 'Bangladesh', 'India', 'Nepal', 'Pakistan',
       'Sri Lanka', 'Other Southern and Central Asia',
       'Total Southern and Central Asia', 'Canada', 'United States of America',
       'Argentina', 'Brazil', 'Chile', 'Colombia', 'Mexico', 'Other Americas',
       'Total Americas', 'South Africa', 'Other Sub-Saharan Africa',
       'Total Sub-Saharan Africa'])
combined_long_df.head()
```

```python
combined_long_df.rename(columns={'variable':'Country/Continent'}, inplace=True)
combined_long_df.head()
```

```python
# Combine with economic data
economic_df.rename(columns={'Country Name':'Country/Continent','year':'Year'}, inplace=True)
combined_long_df["Year"] = combined_long_df["Date"].dt.year
merged_df = combined_long_df.merge(economic_df, on=["Year","Country/Continent"], how="inner")
merged_df.head()
```

```python
merged_df.drop(columns='Unnamed: 0',inplace=True)
```

```python
merged_df.drop_duplicates(inplace=True)
```

```python
merged_df = merged_df.merge(combined_df.loc[:,['Date','Under 1 week', '1 and under 2 weeks',
       '2 weeks and under 1 month', '1 and under 2 months',
       '2 and under 3 months', '3 and under 6 months', '6 and under 12 months',
       'Total (Duration of stay in days)', 'Convention/conference', 'Business',
       'Visiting friends/relatives', 'Holiday', 'Employment', 'Education',
       'Other & not stated', 'Total (Reason for Journey)']], on='Date', how='left').drop_duplicates()
```

```python
merged_df.shape
```

```python
merged_df.drop(columns=['Total (Duration of stay in days)','Total (Reason for Journey)'],inplace=True)
merged_df.rename(columns={'value':'Arrivals'}, inplace=True)
merged_df.columns
```

```python
cols=  ['Under 1 week', '1 and under 2 weeks', '2 weeks and under 1 month', '1 and under 2 months', '2 and under 3 months', '3 and under 6 months', '6 and under 12 months']

# Compute row sums
row_totals = merged_df[cols].sum(axis=1)

# Divide each column by its row total
merged_df[cols] = merged_df[cols].div(row_totals, axis=0)
merged_df.head()
```

```python
cols= ['Convention/conference', 'Business',
       'Visiting friends/relatives', 'Holiday', 'Employment', 'Education',
       'Other & not stated']

# Compute row sums
row_totals = merged_df[cols].sum(axis=1)

# Divide each column by its row total
merged_df[cols] = merged_df[cols].div(row_totals, axis=0)
merged_df.head()
```

```python
new_bitre_df = bitre_df.drop(columns=['AIRLINE','FIN_Y','CAL_Y'])
new_bitre_df.replace('..', np.nan, inplace=True)
new_bitre_df.replace('(d)', np.nan, inplace=True)
new_bitre_df.replace('(e)', np.nan, inplace=True)
new_bitre_df.replace('-', np.nan, inplace=True)
new_bitre_df.replace('(f)', np.nan, inplace=True)
new_bitre_df.replace('(a)', np.nan, inplace=True)
new_bitre_df.replace('(b)', np.nan, inplace=True)
new_bitre_df.replace('(h)', np.nan, inplace=True)
new_bitre_df.replace('(c)', np.nan, inplace=True)
new_bitre_df.replace('(l)', np.nan, inplace=True)
new_bitre_df.replace('(g)', np.nan, inplace=True)
new_bitre_df.replace('(v)', np.nan, inplace=True)
new_bitre_df.replace('(k)', np.nan, inplace=True)
new_bitre_df.replace('(z)', np.nan, inplace=True)
new_bitre_df['PAX IN'] = new_bitre_df['PAX IN'].str.replace(" ", "")
new_bitre_df['SEATS IN'] = new_bitre_df['SEATS IN'].str.replace(" ", "")
new_bitre_df['PAX OUT'] = new_bitre_df['PAX OUT'].str.replace(" ", "")
new_bitre_df['SEATS OUT'] = new_bitre_df['SEATS OUT'].str.replace(" ", "")

#new_bitre_df["COUNTRY"] = new_bitre_df["COUNTRY"].astype('category')
new_bitre_df["FLTS IN"] = new_bitre_df["FLTS IN"].astype('float')
new_bitre_df["PAX IN"] = new_bitre_df["PAX IN"].astype('float')
new_bitre_df["SEATS IN"] = new_bitre_df["SEATS IN"].astype('float')
new_bitre_df["FLTS OUT"] = new_bitre_df["FLTS OUT"].astype('float')
new_bitre_df["PAX OUT"] = new_bitre_df["PAX OUT"].astype('float')
new_bitre_df["SEATS OUT"] = new_bitre_df["SEATS OUT"].astype('float')

new_bitre_df.dtypes
```

```python
new_bitre_df.head()
```

```python
totals = new_bitre_df.groupby(by=["COUNTRY","Date"],as_index=False).sum()
totals.rename(columns={'COUNTRY':'Country/Continent'}, inplace=True)
totals
```
