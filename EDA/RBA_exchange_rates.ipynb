# RBA Exchange Rates

```python
import pandas as pd
```

```python
rba_data_2009 = pd.read_excel("../raw-data/RBA_Exchange_Rates/exchange_rate_1969_2009.xls")
rba_data_2025 = pd.read_excel("../raw-data/RBA_Exchange_Rates/exchange_rate_2010-2025.xls")
```

```python
rba_data_2009.columns = rba_data_2009.iloc[4,:].values
```

```python
rba_data_2009 = rba_data_2009.iloc[10:,:-1]
rba_data_2009.rename(columns={"Units":"Date"},inplace=True)
rba_data_2009["Date"] = pd.to_datetime(rba_data_2009["Date"])
rba_data_2009
```

```python
rba_data_2025
rba_data_2025.columns = rba_data_2025.iloc[4,:].values
rba_data_2025 = rba_data_2025.iloc[10:,:]
rba_data_2025.rename(columns={"Units":"Date"},inplace=True)
rba_data_2025["Date"] = pd.to_datetime(rba_data_2025["Date"])
rba_data_2025
```

```python
rba_data_2009.shape
```

```python
df_merged = pd.concat([rba_data_2009, rba_data_2025], ignore_index=True, sort=False)
df_merged.shape
```

```python
df_merged.to_csv("../processed-csvs/rba_interest_rates.csv")
```
