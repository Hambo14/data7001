# Cleaning Up Short Term Arrivals Data

```python
import pandas as pd

short_term_arrivals = pd.read_excel("../raw-data/merged_output_shorttermarrivals.xlsx")
short_term_arrivals.head()
```

```python
short_term_arrivals.set_index('Date', inplace=True)
short_term_arrivals.head()
```

```python
cleaner_columns = [column.split(';')[1:-1] for column in short_term_arrivals.columns]
#[column.remove('  Short-term Visitors arriving ') for column in cleaner_columns]
cleaned_columns = [(column[0]+column[1]).replace("  Short-term Visitors arriving ", "") for column in cleaner_columns]
#cleaner_columns[30]
short_term_arrivals.columns = cleaned_columns
```

```python
short_term_arrivals.head()
```

```python
short_term_arrivals.to_csv("../processed-csvs/short_term_arrivals.csv")
```
